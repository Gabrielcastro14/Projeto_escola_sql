# Análise 1 — Ranking de Professores por Eficiência

Este notebook tem como objetivo analisar a eficiência dos professores
com base na média de matrículas por aula, utilizando dados provenientes
de um banco de dados relacional.

Os dados analisados são resultados de uma view SQL previamente criada,
garantindo consistência e reaproveitamento das regras de negócio.


```python
from sqlalchemy import create_engine
import pandas as pd

engine = create_engine(
   "postgresql://USUARIO:SENHA@localhost:5432/escola"

df = pd.read_sql(
    "SELECT * FROM vw_ranking_professores;",
    engine
)

df.head()
```

```python
df.plot(
    x="nome_professor",
    y="matriculas_por_aula",
    kind="bar",
    title="Matrículas por Aula por Professor",
    figsize=(10,5)
)
```

### Conclusão

A análise indica que alguns professores apresentam
média significativamente maior de matrículas por aula,
sugerindo maior eficiência ou maior concentração de
demanda em determinadas disciplinas.
